```python
# [Cell 1: Morphometric Terrain Calculation]
import os
import numpy as np
import rasterio
import matplotlib.pyplot as plt

PROCESSED_DIR = os.path.join("..", "data", "processed")
OUTPUTS_DIR = os.path.join("..", "outputs")

dem_path = os.path.join(PROCESSED_DIR, "juja_dem.tif")

with rasterio.open(dem_path) as src:
    dem = src.read(1)
    transform = src.transform

# Calculate surface slope via gradient
dy, dx = np.gradient(dem, transform[0], -transform[4])
slope_rad = np.arctan(np.sqrt(dx**2 + dy**2))
slope_deg = np.degrees(slope_rad)

# [Cell 2: Visualize Slope & Export Map]
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

im0 = ax[0].imshow(dem, cmap='magma')
ax[0].set_title("Elevation (m)")
plt.colorbar(im0, ax=ax[0])

im1 = ax[1].imshow(slope_deg, cmap='YlOrRd')
ax[1].set_title("Slope Angle (Degrees)")
plt.colorbar(im1, ax=ax[1])

map_out = os.path.join(OUTPUTS_DIR, "maps", "juja_slope_analysis.png")
plt.savefig(map_out, dpi=300, bbox_inches='tight')
plt.show()
print(f"Exported terrain analysis to {map_out}")
```
