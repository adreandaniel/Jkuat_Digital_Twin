```python
# [Cell 1: Imports & Time-Series Parsing]
import os
import pandas as pd
import matplotlib.pyplot as plt

PROCESSED_DIR = os.path.join("..", "data", "processed")
OUTPUTS_DIR = os.path.join("..", "outputs")

parquet_path = os.path.join(PROCESSED_DIR, "rainfall_events.parquet")
df_rain = pd.read_parquet(parquet_path)
df_rain['timestamp'] = pd.to_datetime(df_rain['timestamp'])

print("--- Rainfall Event Summary ---")
print(df_rain.describe())

# [Cell 2: Plot Storm Intensity Hydrograph]
plt.figure(figsize=(10, 5))
plt.bar(df_rain['event_id'], df_rain['intensity_mm_hr'], color='skyblue', edgecolor='navy')
plt.axhline(30.0, color='red', linestyle='--', label='Flood Threshold (30 mm/hr)')
plt.ylabel("Rainfall Intensity (mm/hr)")
plt.title("Storm Event Intensity Comparison")
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.7)

fig_path = os.path.join(OUTPUTS_DIR, "figures", "rainfall_timeseries.png")
plt.savefig(fig_path, dpi=300, bbox_inches='tight')
plt.show()
print(f"Saved figure to {fig_path}")
```
