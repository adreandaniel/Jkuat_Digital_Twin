```python
# [Cell 1: Imports & Setup]
import os
import json
import numpy as np
import pandas as pd
import geopandas as gpd
import rasterio
import matplotlib.pyplot as plt

DATA_DIR = os.path.join("..", "data")
PROCESSED_DIR = os.path.join(DATA_DIR, "processed")
OUTPUTS_DIR = os.path.join("..", "outputs")

# [Cell 2: Inspect Raster Assets]
dem_path = os.path.join(PROCESSED_DIR, "juja_dem.tif")
with rasterio.open(dem_path) as src:
    print(f"--- DEM Raster Metadata ---")
    print(f"Driver: {src.driver}")
    print(f"Dimensions: {src.width} x {src.height} cells")
    print(f"Bands: {src.count}")
    print(f"CRS: {src.crs}")
    print(f"Bounds: {src.bounds}")
    dem_array = src.read(1)

plt.figure(figsize=(8, 6))
plt.imshow(dem_array, cmap='terrain')
plt.colorbar(label='Elevation (m)')
plt.title("Juja Digital Elevation Model (DEM)")
plt.savefig(os.path.join(OUTPUTS_DIR, "maps", "juja_elevation_map.png"), dpi=300)
plt.show()

# [Cell 3: Inspect Vector Assets]
streams_path = os.path.join(PROCESSED_DIR, "juja_streams.geojson")
streams_gdf = gpd.read_file(streams_path)
print("\n--- Stream Vector Metadata ---")
print(f"Feature Count: {len(streams_gdf)}")
print(f"CRS: {streams_gdf.crs}")
print(streams_gdf.head())
```
